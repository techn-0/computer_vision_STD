```python
import skimage
import numpy as np
import cv2 as cv
from google.colab.patches import cv2_imshow

img = skimage.data.coffee()
cv2_imshow(cv.cvtColor(img, cv.COLOR_RGB2BGR))
# cv.imshow('Coffee image', cv.cvtColor(img, cv.COLOR_RGB2BGR))

slic1 = skimage.segmentation.slic(img, compactness= 30, n_segments= 500)
sp_img1 = skimage.segmentation.mark_boundaries(img, slic1)
sp_img1 = np.uint8(sp_img1*255.0)

slic2 = skimage.segmentation.slic(img, compactness= 90, n_segments= 800)
sp_img2 = skimage.segmentation.mark_boundaries(img, slic2)
sp_img2 = np.uint8(sp_img2*255.0)

cv2_imshow(cv.cvtColor(sp_img1, cv.COLOR_RGB2BGR))
cv2_imshow(cv.cvtColor(sp_img2, cv.COLOR_RGB2BGR))

# cv.imshow('Super pixels (compact 100)', cv.cvtColor(sp_img1, cv.COLOR_RGB2BGR))
# cv.imshow('Super pixels (compact 200)', cv.cvtColor(sp_img2, cv.COLOR_RGB2BGR))
# cv.waitKey(0)
# cv.destroyAllWindows()
```
