```python
import cv2 as cv

img = cv.imread('')

gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

canny1 = cv.Canny(gray, 30, 100)
canny2 = cv.Canny(gray, 70, 180)

cv.imshow('gray', gray)
cv.imshow('canny1', canny1)
cv.imshow('canny2', canny2)

# cv.waitKey()
# cv.destroyAllWindows()
```
